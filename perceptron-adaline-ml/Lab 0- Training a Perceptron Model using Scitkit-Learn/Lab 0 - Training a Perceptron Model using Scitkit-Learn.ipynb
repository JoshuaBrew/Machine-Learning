<h2 style = "color:blue">Let's import the iris flower dataset that is packaged with Scitkit-Learn</h2>

### Choosing a dataset and classificantion algorithm

#### Let's load the scitkit-learn datasets and import the iris flower dataset

Loading the iris dataset from scitkit-learn. Here, the third column represents the petal length, and the fourth column the petal width of the flower samples. The classes have already converted to integer labels with: <b>0=Iris-Setosa, 1=Iris-Versicolor, 2=Iris-Virginica.</b>

Note that we are using the entire class labels with 150 flowers but we will restrict our dataset to two columns for illustrative purposes.

```python
from sklearn import datasets
import numpy as np

iris = datasets.load_iris()
X = iris.data[:,[2,3]]
y = iris.target

print('Class labels:', np.unique(y))

# importing a sklearn dataset
# importing a numerical tool in numpy
# setting collumns of the dataset to X and y variables
# printing the class labels
```

### Now, let's split the data into 75% training and 25% test data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0)

# importing a split operator and splitting the pulled data in X and y
# tests on 25% of the data are done to train the model
```

<h2 style="color: blue;">We need to scale (normalize) our feature matrix:</h2>

### Feature Scaling
    
<h2 style="color:red;">IMPORTANT: Standardizing the Features:</h2>

***Feature Scaling***(https://en.wikipedia.org/wiki/Feature_scaling)

One of the most important transformations you need to apply to your data is feature scaling. With very few exceptions, machine learning algorithms will not perform well when the input numerical attributes have very different scales(we first encountered this when using Euclidean distances for classification). In Scitkit-Learn, standardization is usually a requirement for many machine learning alogrithms. Models that use datasets that do not look more or less like they are standard-normally distributd will behave badly (show poor performance, not converge, etc). This means that each column in the feature dataset should have a Caussian (Normal) distribution with zero mean and unit variance. Scitkit-Learn has a built-in tool for performing this transformation which is called ***StandardScaler***.

#### Let's use the standardScaler from Scitkit-learn to transform (scale) our feature dataset

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
sc.fit(X_train)
X_train_std = sc.transform(X_train)
X_test_std = sc.transform(X_test)

# importing a StandardScaler and setting it to a sc variable
# fitting the sc data to X_train
# Transforming the dataset by scaling it
```

<h3 style="color:red;">Pay close attention to the above steps used to standardizing the features of the dataset</h3>

<i>Also note that once standardized, we use the transformed dataset for the training, etc.</i>

<h2 style="color:red;">We are now ready to train the model:</h2>

### Training a Perceptron Model using Scitkit-Learn

#### We are now ready to train the perceptron model on the transformed feature data

```python
from sklearn.linear_model import Perceptron

ppn = Perceptron(max_iter=40, eta0=0.1, tol = 0.01, random_state=0)
ppn.fit(X_train_std, y_train)

# ppn are the parameters the perceptron is set to
# max_iter shows the number of iterations
# eta0 is the learning rate
# tol determines when the model stops training
```

### Shape of data

```python
y_test.shape

# the shape of the y set
```

### Misclassifications

```python
y_pred = ppn.predict(X_test_std)
print("Misclassified samples: {0}".format((y_test != y_pred).sum()))

# Displaying the number of misclassifications
```

### Accuracy

```python
from sklearn.metrics import accuracy_score

print("Accurancy: {0:.2f}".format(accuracy_score(y_test, y_pred)))
print("\nThe accuracy score of the model is: {0:0.2f}".format(ppn.score(X_test_std,y_test)))

# Printing the accuracy score with the import from sklearn.metrics
```
