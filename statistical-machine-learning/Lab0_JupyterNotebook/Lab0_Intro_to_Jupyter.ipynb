## Prime of numbers

Let's now look at a function that checks whether a number is prime

```python
def is_prime_naive(n):
    for i in range(2,n):         # Note: range(n, m) ->, na+1, ..., m-1
        if n % i ==0:            # % is the modulo operation.
            return False
    return True
```

Let's try it out by calling the function:

```python
is_prime_naive(2**19 - 1), is_prime_naive(1000000)
```

## Indexing and Slicing

Pyhton index lists begin at zero and the NumPy arrays follow suit. When indexing lists in Python, we
normally do the following for a 2 x 2 object:

```python
alist=[
        [1,2],
        [3,4]
     ]
```

```python
# To return the (0,1) element we must index as shown below.
alist[0][1]
```

```python
#In NumPy, indexing follows a more convenient syntax.
# Converting the list defined above into an array
arr = np.array(alist)
# To return the (0,1) element we use...
arr[0,1]
# Now to acces the last column, we simply use...
print(arr[:,1])
# Accessing the rows is achieved in the same way,
arr[1,:]
# which is the bottom row.
```

### Linear Algebra

*NumPy arrays do not behave like matrices in linear algebra by default. Instead, the operations are
mapped from each element in one array onto the next. This is quite a useful feature, as loop
operations can be done away with for efficiency. But what about when transposing or a dot
multiplication are needed? Without invoking other classes, you can use the built-in numpy.dot and
numpy.transpose to do such operations. The syntax is Pythonic, so it is intuitive to program. Or the
math purist can use the numpy.matrix object instead. We will go over both examples below to
illustratethe differences and similarities between the two options. More importantly, we will compare
some of the advantages and disadvantages between the numpy.array and the numpy.matrix objects
Some operations are easy and quick to do in linear algebra. A classic example is solving a system
of equations that we can express in matrix form:*

*Now let us represent the matrix system as AX = B, and solve for the variables. This means we
should try to obtain X = A(-1)B. Here is how we would do this with NumPy.*

```python
import numpy as np
# Defining the matrices
A = np.matrix([[3, 6, -5],
[1, -3, 2],
[5, -1, 4]])
B = np.matrix([[12],
[-2],
[10]])
# Solving for the variables, where we invert A
X = A ** (-1) * B
print(X)
```

### Another Way

Now that we understand how NumPy matrices work, we can show how to do the same operations
without specifically using the numpy.matrix subclass. (The numpy.matrix subclass is contained
within the numpy.array class, which means that we can do the same example as that above without
directly invoking the numpy.matrix class.)

```python
import numpy as np 
a=np.array([[3, 6, -5],
[1, -3, 2],
[5, -1,4]])
# Defining the array 
b = np.array([12,-2, 10])
# Solving for the variables, where we invert A
x = np.linalg.inv(a).dot(b)
print(x)
```

### Using MatPlotLib

The matplotlib package is the standard way to plot data in Python. To use MatPlotLib in our
notebook we have to use the *magic command*

```python
# allow plots to apprear within the notebook
%matplotlib inline
# import MatPlotlib (scientific plottting library)
import matplotlib.pyplot as plt
```

**Let's use *list* comprehensions to create data for a simple plot**

```python
import math
x_data = [0.05*x for x in range(101)]
y_data = [5*math.sin(2*math.pi*x) for x in x_data]
fig = plt.figure(figsize=(4,4))
plt.plot(x_data,y_data,'g--')
plt.show()
fig.savefig('Images/sine_wave444.pdf', dpi=300)
```

Let's do the same thing with **numpy**

```python
# generate x-y data
x = np.linspace(0,5,100)
y = 5*np.sin(2*np.pi*x)
# ser the plot size
fig = plt.figure(figsize=(4,4))
# plot data and show the plot
plt.plot(x,y)
plt.show()
```

### Descriptive Statistics and Pandas Dataframe

```python
import pandas as pd
```

Pandas introduces an object called a **DataFrame**. A **DataFrame** is a tabular data structure
compresed of rows and columns, like those in a spreadsheet or a database table.

Let's create some data to use in this tutorial.

```python
data = {'name': ['Jason', 'Molly', 'Tine', 'Jake', 'Amy'],
       'age': [20, 21, 24, 27, 33],
       'preTestScore': [70, 70, 67, 68, 78],
       'postTestScore': [70, 82, 69, 74, 86]}
df = pd.DataFrame(data, columns = ['name', 'age', 'preTestScore', 'postTestScore'])
```
