```python
import matplotlib.pyplot as plt
import numpy as np
import math

# Here we imported packages
```

```python
X = np.linspace(-5, 5, 200)
Z = 1/(1+np.exp(-X))
def sigmoid(z):
    return 1/(1+np.exp(-z))

# numpy an array to fit the sigmoid function
# Sigmoid makes it stochastic
```

```python
plt.plot(X,Z)
plt.xlabel("Z")
plt.ylabel("Y")
plt.title("Sigmoid Function")

plt.show
```

```python
proba = sigmoid(0)
print('Probability at z=0:', proba)
```
